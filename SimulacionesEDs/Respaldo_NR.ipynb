# **Respaldo de método gráfico de solución: Newton-Raphson**

```python
import sympy as sp
import numpy as np
```

*Configuración de ecuación a resolver en forma de función*

$f(x)=\frac{4lu_0^2\theta_0^4x}{gx^4+3lu_0^2\theta_0^4}-\bar{\theta} \Rightarrow \frac{4lu_0^2\theta_0^4x}{gx^4+3lu_0^2\theta_0^4}-\bar{\theta}=0$ (encontrar el cero positivo de la función)

Constantes

```python
l, g = 0.270, 9.8
```

Primer conjunto de condiciones iniciales

```python
u0, theta0, thetaf = 5.35, np.pi / 12, 0.232
thetaPromedio = (theta0 + thetaf) / 2
```

```python
x = sp.symbols('x')

f = (4 * l * u0**2 * theta0**4 * x) / (g * x**4 + 3 * l * u0**2 * theta0**4) - thetaPromedio

df = sp.diff(f, x)

# Para convertir f y df en funciones numéricas
f = sp.lambdify(x, f, "numpy")
df = sp.lambdify(x, df, "numpy")
```

*Función del método Newton-Raphson*

```python
def newton_raphson(x0, tolerancia=1e-8, iter_max=500):
    # Valor inicial
    xn = x0

    # Ciclo asociado a proceso iterativo
    for i in range(iter_max):
        if df(xn)==0: raise ZeroDivisionError("La derivada es nula. No se encontró una solución.")
        # Ejecución de la expresión del método
        xn_1 = xn - f(xn) / df(xn)

        # Verificación de tolerancia
        if abs(xn_1-xn)<tolerancia: return round(xn_1, 3)
        # Validación de divergencia de la solución (si el próximo valor representa un cambio brusco respecto a los soluciones convergentes, entonces regresa el valor actual)
        elif abs((xn_1 - f(xn_1) / df(xn_1))-xn_1) / abs(xn_1-xn) > 7: return round(xn_1, 3)

        # Actualización del valor xn
        xn = xn_1
    
    raise RuntimeError("Número máximo de iteraciones alcanzada. No se encontró una solución.")
```

```python
theta_M = newton_raphson(0)
print(f"Solución encontrada: {theta_M}")
print(f"theta promedio: {round(thetaPromedio, 3)}")
print(f"Derivada evaluada en theta_M: {df(theta_M)}")
```
